```python
# ============================================================
# IMPORTACIÓN DE LIBRERÍAS
# ============================================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.impute import SimpleImputer

import ipywidgets as widgets
from IPython.display import display, clear_output


# ============================================================
# CARGAR DATASET
# ============================================================

df = pd.read_csv("churn_limpio.csv", sep = ";")

print("Dataset cargado correctamente")
print(f"Filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")

display(df.head())


# ============================================================
# FUNCIÓN 1: ELIMINACIÓN DE VALORES NULOS
# ============================================================

def eliminar_nulos(df):
    """Elimina filas con valores nulos."""
    df_sin_nulos = df.dropna()
    return df_sin_nulos


# ============================================================
# FUNCIÓN 2: CODIFICACIÓN DE VARIABLES CATEGÓRICAS
# ============================================================

def codificar_categoricas(df, columnas):
    """Codifica columnas categóricas usando LabelEncoder."""

    df_codificado = df.copy()

    for col in columnas:
        le = LabelEncoder()
        df_codificado[col] = le.fit_transform(
            df_codificado[col].astype(str)
        )

    return df_codificado


# ============================================================
# FUNCIÓN 3: NORMALIZACIÓN DE DATOS NUMÉRICOS
# ============================================================

def normalizar_datos(df, columnas):
    """Normaliza columnas numéricas entre 0 y 1."""

    df_normalizado = df.copy()

    scaler = MinMaxScaler()

    df_normalizado[columnas] = scaler.fit_transform(
        df_normalizado[columnas]
    )

    return df_normalizado


# ============================================================
# IDENTIFICAR COLUMNAS
# ============================================================

columnas_categoricas = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

columnas_numericas = df.select_dtypes(
    include=np.number
).columns.tolist()


print("\nColumnas categóricas:")
print(columnas_categoricas)

print("\nColumnas numéricas:")
print(columnas_numericas)


# ============================================================
# WIDGETS
# ============================================================

widget_categoricas = widgets.SelectMultiple(
    options=columnas_categoricas,
    description="Categóricas:",
    layout=widgets.Layout(width="500px", height="150px")
)

widget_numericas = widgets.SelectMultiple(
    options=columnas_numericas,
    description="Numéricas:",
    layout=widgets.Layout(width="500px", height="150px")
)

boton_procesar = widgets.Button(
    description="Procesar dataset",
    button_style="success",
    icon="check"
)

salida = widgets.Output()


# ============================================================
# FUNCIÓN PRINCIPAL
# ============================================================

def procesar_dataset(boton):

    with salida:

        clear_output()

        # Obtener columnas seleccionadas
        cols_categoricas = list(widget_categoricas.value)
        cols_numericas = list(widget_numericas.value)

        print("Columnas categóricas seleccionadas:")
        print(cols_categoricas)

        print("\nColumnas numéricas seleccionadas:")
        print(cols_numericas)

        # ----------------------------------------------------
        # 1. Eliminar valores nulos
        # ----------------------------------------------------

        df_procesado = eliminar_nulos(df)

        print("\nDespués de eliminar valores nulos:")
        print(df_procesado.shape)

        # ----------------------------------------------------
        # 2. Codificar variables categóricas
        # ----------------------------------------------------

        if cols_categoricas:

            df_procesado = codificar_categoricas(
                df_procesado,
                cols_categoricas
            )

            print("\nVariables categóricas codificadas.")

        # ----------------------------------------------------
        # 3. Normalizar variables numéricas
        # ----------------------------------------------------

        if cols_numericas:

            df_procesado = normalizar_datos(
                df_procesado,
                cols_numericas
            )

            print("Variables numéricas normalizadas.")

        # ----------------------------------------------------
        # 4. Crear dataset final
        # ----------------------------------------------------

        churn_normalizado = df_procesado.copy()

        # ----------------------------------------------------
        # 5. Guardar CSV
        # ----------------------------------------------------

        churn_normalizado.to_csv(
            "churn_normalizado.csv",
            index=False
        )

        print("\n" + "=" * 50)
        print("PROCESO COMPLETADO")
        print("=" * 50)

        print("\nDataset generado:")
        print("churn_normalizado.csv")

        print("\nDimensiones:")
        print(churn_normalizado.shape)

        print("\nPrimeras filas:")

        display(churn_normalizado.head())


# ============================================================
# CONECTAR BOTÓN CON FUNCIÓN
# ============================================================

boton_procesar.on_click(procesar_dataset)


# ============================================================
# MOSTRAR INTERFAZ
# ============================================================

display(
    widgets.VBox([
        widgets.HTML("<h3>Configuración del procesamiento</h3>"),

        widgets.HTML(
            "<b>Selecciona las variables categóricas que deseas codificar:</b>"
        ),

        widget_categoricas,

        widgets.HTML(
            "<b>Selecciona las variables numéricas que deseas normalizar:</b>"
        ),

        widget_numericas,

        boton_procesar,

        salida
    ])
)
```
